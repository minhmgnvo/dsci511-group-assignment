__3a.__ (10 points) Define a function called `get_parsed_page` that takes the title of a page as input and uses the Wikipedia REST API ([documentation here](https://en.wikipedia.org/api/rest_v1/#/)) to return the HTML content on this page. Use the `requests` module to access the page, and use `BeautifulSoup` to parse the API response. The function should return the original response and the parsed response. Apply `get_parsed_page` to "List_of_cities_in_the_United_Kingdom", and print the content stored under the 'head' tag.

````python
import requests
from bs4 import BeautifulSoup

def get_parsed_page(page_title):
    """Fetches the content of a Wikipedia page using the REST API."""
    url = f"https://en.wikipedia.org/api/rest_v1/page/html/{page_title}"
    res = requests.get(url=url, headers={"User-Agent": "Mozilla/5.0"})
    doc = res.text
    soup = BeautifulSoup(doc, "html.parser")
    return soup, doc


page_soup, page_response = get_parsed_page("List_of_cities_in_the_United_Kingdom")
page_soup.find('head').text
````

__3b.__ (15 points) Locate the first object tagged `'table'` in the parsed HTML output above. Print the length of the list and the first five items in `city_links`.

````python
city_links = []
city_table = page_soup.find('table')
table = city_table
#---Your code starts here---
rows = table.find_all('tr')[1:]  
for row in rows:
    first_col = row.find('td') 
    if first_col:
        a_tag = first_col.find('a')
        if a_tag:
            text = a_tag.text
            link = a_tag.get('href')
            city_links.append((text, link))


#---Your code stops here---

print("Length of list:", len(city_links), "\n")
city_links[:5]
````
